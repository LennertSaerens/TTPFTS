```python
import matplotlib.pyplot as plt

from plotting import plot_metric_all_envs_grid_top_legend
import os
```

```python
num_runs = 100
budget = 5001
DATA_DIR = "results"
FIGURES_DIR = "figures"
```

```python
plt.rcParams.update({'font.size': 14, 'font.family': 'serif'})
```

```python
env_files = []
env_labels = []
env_arms = [60, 10, 200, 50, 20, 10, 22, 5]
env_objs = [2, 2, 2, 10, 2, 2, 2, 2]

# EgeExp1–8
for i in range(1, 9):
    f = f"{DATA_DIR}/EgeExp{i}_main_combined.csv"
    if os.path.exists(f):
        env_files.append(f)
        env_labels.append(f"EgeExp{i}")
```

```python
psi_metrics = ["Jaccard", "Bernoulli", "Misclassification"]

for metric in psi_metrics:
    plot_metric_all_envs_grid_top_legend(
        env_files,
        env_labels,
        env_arms,
        env_objs,
        metric=metric,
        num_runs=num_runs,
        num_arm_pulls=budget,
        plot_std=True,
        step=1,
        save_png=True,
        save_file=f"{FIGURES_DIR}/{metric.lower()}_all_envs.png"
    )
```
